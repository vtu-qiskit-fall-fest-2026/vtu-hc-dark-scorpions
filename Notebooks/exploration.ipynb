```python
# QAOA Exploration — Dark Scorpions
import networkx as nx
import numpy as np
import matplotlib.pyplot as plt
from src.classical import classical_maxcut
from src.optimizer import quantum_maxcut
```

```python
G = nx.complete_graph(4)
print(f"Nodes: {G.number_of_nodes()}, Edges: {G.number_of_edges()}")
```

```python
c_cut, c_part = classical_maxcut(G, 4)
print(f"Classical: cut={c_cut}, partition={c_part}")   
```

```python
q_cut, q_part, q_energy = quantum_maxcut(G, 4, reps=1)
print(f"QAOA p=1: cut={q_cut}, energy={q_energy:.4f}")   
```

```python
q_cut2, q_part2, q_energy2 = quantum_maxcut(G, 4, reps=2)
print(f"QAOA p=2: cut={q_cut2}, energy={q_energy2:.4f}")
```

```python
cuts_c, cuts_q, densities = [], [], [0.3, 0.4, 0.5, 0.6, 0.7, 0.8]
for d in densities:
    G = nx.erdos_renyi_graph(4, d, seed=42)   
    c, _ = classical_maxcut(G, 4)
    q, _, _ = quantum_maxcut(G, 4, reps=2)
    cuts_c.append(c); cuts_q.append(q)
    print(f"d={d}: classical={c}, qaoa={q}")

plt.plot(densities, cuts_c, 'bo-', label='Classical')
plt.plot(densities, cuts_q, 'rs-', label='QAOA p=2')
plt.xlabel('Edge density'); plt.ylabel('Max Cut'); plt.legend()
plt.title('MaxCut vs Edge Density (4 nodes)'); plt.show()   
```
